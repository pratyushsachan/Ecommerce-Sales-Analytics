```python
df_orders_delivered = df_orders[df_orders['order_status'] == 'delivered'].copy()
print(df_orders_delivered.shape)
```

```python
# Start with delivered orders + customers
df_master = df_orders_delivered.merge(df_customers, on='customer_id', how='left')

# Add order items (this creates one row per item, so order-level rows will repeat if an order has multiple items)
df_master = df_master.merge(df_order_items, on='order_id', how='left')

# Add products
df_master = df_master.merge(df_products, on='product_id', how='left')

# Add product category translation (Portuguese -> English)
df_master = df_master.merge(df_category_translation, on='product_category_name', how='left')

# Add sellers
df_master = df_master.merge(df_sellers, on='seller_id', how='left')

# Add payments
df_master = df_master.merge(df_payments, on='order_id', how='left')

# Add reviews
df_master = df_master.merge(df_reviews, on='order_id', how='left')

print(df_master.shape)
df_master.head()
```

```python
print("Original delivered orders:", df_orders_delivered.shape[0])
print("Master dataset rows:", df_master.shape[0])
print("Unique orders in master:", df_master['order_id'].nunique())
```

```python
# Convert date columns from text to actual datetime objects so we can do date math on them
df_master['order_delivered_customer_date'] = pd.to_datetime(df_master['order_delivered_customer_date'])
df_master['order_estimated_delivery_date'] = pd.to_datetime(df_master['order_estimated_delivery_date'])
df_master['order_purchase_timestamp'] = pd.to_datetime(df_master['order_purchase_timestamp'])

# Calculate delivery delay in days: actual delivery date minus estimated delivery date
# Positive number = delivered late, negative number = delivered early
df_master['delivery_delay_days'] = (
    df_master['order_delivered_customer_date'] - df_master['order_estimated_delivery_date']
).dt.days

# Quick check: see the distribution of delays (min, max, average)
print(df_master['delivery_delay_days'].describe())
```

```python
# Extract year and month from purchase date so we can group revenue/orders by month later
df_master['order_year'] = df_master['order_purchase_timestamp'].dt.year
df_master['order_month'] = df_master['order_purchase_timestamp'].dt.month

# Create a combined "year-month" column, useful for plotting monthly trends in Power BI
df_master['order_year_month'] = df_master['order_purchase_timestamp'].dt.to_period('M').astype(str)
```

```python
# Total price paid for this specific item = product price + freight (shipping) cost
# This is the item-level revenue, useful for category/product-level analysis
df_master['item_revenue'] = df_master['price'] + df_master['freight_value']
```

```python
# Create the processed folder if it doesn't exist, using path relative to repo root
os.makedirs('data/processed', exist_ok=True)

# Save using the correct relative path (no ../ needed since we're already at repo root)
df_master.to_csv('data/processed/olist_master_v1.csv', index=False)
print("Saved master dataset with", df_master.shape[0], "rows")
```

```python
# Recency needs a reference "today" date — since this is historical data, 
# we use the day after the last purchase in the dataset as our "today"
reference_date = df_master['order_purchase_timestamp'].max() + pd.Timedelta(days=1)

# Group by customer_unique_id (not customer_id, since Olist creates a new customer_id per order,
# but customer_unique_id stays the same for a repeat customer)
rfm = df_master.groupby('customer_unique_id').agg(
    # Recency: days since their most recent purchase (smaller = more recent = better)
    recency=('order_purchase_timestamp', lambda x: (reference_date - x.max()).days),
    
    # Frequency: number of unique orders this customer placed
    frequency=('order_id', 'nunique'),
    
    # Monetary: total amount this customer has spent across all orders
    monetary=('item_revenue', 'sum')
).reset_index()

# Quick look at the result
print(rfm.shape)
rfm.head()
```

```python
# qcut splits customers into 5 equal-sized buckets (quintiles) based on each metric
# For Recency: LOWER days = better, so we reverse the labels (5 = most recent)
rfm['r_score'] = pd.qcut(rfm['recency'], 5, labels=[5, 4, 3, 2, 1])

# For Frequency: HIGHER count = better, so normal order (5 = most frequent)
# rank(method='first') handles ties (many customers have frequency=1) so qcut doesn't error
rfm['f_score'] = pd.qcut(rfm['frequency'].rank(method='first'), 5, labels=[1, 2, 3, 4, 5])

# For Monetary: HIGHER spend = better (5 = highest spender)
rfm['m_score'] = pd.qcut(rfm['monetary'], 5, labels=[1, 2, 3, 4, 5])

# Combine into a single RFM score string, e.g. "545" = very recent, frequent, high spender
rfm['rfm_score'] = rfm['r_score'].astype(str) + rfm['f_score'].astype(str) + rfm['m_score'].astype(str)

rfm.head()
```

```python
# Define a function that labels each customer based on their R and F scores
# This is a simplified, commonly-used segmentation logic
def segment_customer(row):
    r = int(row['r_score'])
    f = int(row['f_score'])
    
    if r >= 4 and f >= 4:
        return 'Champions'          # Bought recently, buys often, spends well
    elif r >= 3 and f >= 3:
        return 'Loyal Customers'    # Good recency and frequency
    elif r >= 4 and f <= 2:
        return 'New Customers'      # Recent but low frequency (just started buying)
    elif r <= 2 and f >= 3:
        return 'At Risk'            # Used to buy often, but haven't recently — losing them
    elif r <= 2 and f <= 2:
        return 'Lost'               # Low recency and frequency — likely churned
    else:
        return 'Needs Attention'    # Doesn't clearly fit other buckets

# Apply the function to every row
rfm['segment'] = rfm.apply(segment_customer, axis=1)

# See how many customers fall into each segment
print(rfm['segment'].value_counts())
```

```python
# Filter to only rows that have both a delivery delay value and a review score
# (some orders might be missing one or the other, so we drop those for this specific analysis)
delay_review = df_master.dropna(subset=['delivery_delay_days', 'review_score']).copy()

# Create a simple category: was the order late or on-time/early?
# delivery_delay_days > 0 means it arrived AFTER the estimated date (late)
delay_review['delivery_status'] = delay_review['delivery_delay_days'].apply(
    lambda x: 'Late' if x > 0 else 'On-time/Early'
)

# Compare average review score for late vs on-time orders
avg_review_by_status = delay_review.groupby('delivery_status')['review_score'].agg(['mean', 'count'])
print(avg_review_by_status)
```

```python
# Create delay buckets to see if review score drops progressively with longer delays
def delay_bucket(days):
    if days <= 0:
        return '1. On-time/Early'
    elif days <= 3:
        return '2. Late 1-3 days'
    elif days <= 7:
        return '3. Late 4-7 days'
    else:
        return '4. Late 8+ days'

# Apply the bucketing function to every row
delay_review['delay_bucket'] = delay_review['delivery_delay_days'].apply(delay_bucket)

# Average review score per bucket, sorted by the bucket order (not alphabetically)
bucket_summary = delay_review.groupby('delay_bucket')['review_score'].agg(['mean', 'count']).sort_index()
print(bucket_summary)
```

```python
# Save the delay-review summary so you can pull it directly into Power BI or reference it in your README
bucket_summary.to_csv('data/processed/delay_vs_review_summary.csv')
print("Saved delay vs review summary")
```

```python
import sqlite3

# Create (or connect to) a SQLite database file inside your project
# This creates 'ecommerce.db' in your repo root if it doesn't already exist
conn = sqlite3.connect('ecommerce.db')

# Write the master dataframe into a SQL table called 'orders_master'
# if_exists='replace' means if you run this again, it overwrites the old table instead of duplicating rows
df_master.to_sql('orders_master', conn, if_exists='replace', index=False)

# Also load the RFM table as a separate SQL table, since it's a different granularity (customer-level)
rfm.to_sql('customer_rfm', conn, if_exists='replace', index=False)

print("Data loaded into SQLite database: ecommerce.db")
```

```python
# Write a simple SQL query as a Python string
query = """
SELECT order_status, COUNT(DISTINCT order_id) as total_orders
FROM orders_master
GROUP BY order_status;
"""

# Run the query and load results directly into a pandas dataframe
result = pd.read_sql_query(query, conn)
print(result)
```

```python
query = """
-- Find total revenue and order count per product category, ranked highest to lowest
-- Purpose: Identify which categories drive the most revenue
SELECT 
    product_category_name_english as category,
    COUNT(DISTINCT order_id) as total_orders,
    ROUND(SUM(item_revenue), 2) as total_revenue
FROM orders_master
WHERE product_category_name_english IS NOT NULL
GROUP BY product_category_name_english
ORDER BY total_revenue DESC
LIMIT 10;
"""
result = pd.read_sql_query(query, conn)
print(result)
```

```python
query = """
-- Track total revenue by month to see growth/decline over time
-- Purpose: Identify seasonality or growth trends
SELECT 
    order_year_month,
    ROUND(SUM(item_revenue), 2) as monthly_revenue,
    COUNT(DISTINCT order_id) as total_orders
FROM orders_master
GROUP BY order_year_month
ORDER BY order_year_month;
"""
result = pd.read_sql_query(query, conn)
print(result)
```

```python
query = """
-- Find the top 10 sellers generating the most revenue
-- Purpose: Identify high-performing sellers worth partnering closer with
SELECT 
    seller_id,
    seller_state,
    COUNT(DISTINCT order_id) as total_orders,
    ROUND(SUM(item_revenue), 2) as total_revenue
FROM orders_master
GROUP BY seller_id, seller_state
ORDER BY total_revenue DESC
LIMIT 10;
"""
result = pd.read_sql_query(query, conn)
print(result)
```

```python
query = """
-- Total revenue and order count grouped by customer's state
-- Purpose: Identify which regions to prioritize for marketing/logistics
SELECT 
    customer_state,
    COUNT(DISTINCT order_id) as total_orders,
    ROUND(SUM(item_revenue), 2) as total_revenue
FROM orders_master
GROUP BY customer_state
ORDER BY total_revenue DESC;
"""
result = pd.read_sql_query(query, conn)
print(result)
```

```python
query = """
-- Revenue and customer count by RFM segment
-- Purpose: Quantify the size and value of each customer segment
SELECT 
    segment,
    COUNT(*) as customer_count,
    ROUND(SUM(monetary), 2) as total_revenue,
    ROUND(AVG(monetary), 2) as avg_revenue_per_customer
FROM customer_rfm
GROUP BY segment
ORDER BY total_revenue DESC;
"""
result = pd.read_sql_query(query, conn)
print(result)
```
